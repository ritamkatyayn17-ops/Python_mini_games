```python
def mad_libs():
  print("\nlet's create a story in fill in the blanks.")

  noun = input("noun: ")
  verb = input("verb: ")
  adjective = input("adjective:")
  place = input("place:")#formatted string example and multiline comment

  story = f"""
  Once upon a time, a {adjective} {noun}
  decided to {verb} to the {place}.
  Everyone was surprised, but it turned out to be the best idea ever!
  """

  print("\nHere's your story:\n")
  print(story)

if __name__ == "__main__":
  mad_libs()
```

```python
#rock, paper, scissors using random module
import random
choices = ["rock", "paper", "scissors"]#list

print("Welcome to Rock, Paper, Scissors! Type 'quit' to end the game.\n")

while True:
    player = input("choose rock, paper, or scissors (or 'quit' to exit): ").lower()
    if player == 'quit':
        print("Thanks for playing!")
        break
    if player not in choices:
        print("Invalid choice. Please choose rock, paper, or scissors.")
        continue

    computer = random.choice(choices)
    if player == computer:
        print(f"It's a tie! Both chose {player}.\n")
    elif (
        (player == "rock" and computer == "scissors") or
        (player == "paper" and computer == "rock") or
        (player == "scissors" and computer == "paper")
    ):#relational operators
        print(f"You win! {player} beats {computer}.\n")
    else:
        print(f"You lose! {computer} beats {player}.\n")
```

```python
#learning capitals of different countries
questions = {
    "what is the capital of France?": "Paris",
    "what is the capital of Germany?": "Berlin",
    "what is the capital of Italy?": "Rome",
    "what is the capital of Japan?": "Tokyo",
    "what is the capital of the United States?": "Washington, D.C.",
    "what is the capital of Canada?": "Ottawa",
    "what is the capital of Australia?": "Canberra",
    "what is the capital of Brazil?": "Brasília",
    "what is the capital of India?": "New Delhi",
    "what is the capital of China?": "Beijing",
    "what is the capital of Russia?": "Moscow",
    "what is the capital of South Korea?": "Seoul",
    "what is the capital of Mexico?": "Mexico City",
    "what is the capital of Argentina?": "Buenos Aires",
    "what is the capital of Switzerland?": "Bern",
}

def ask_questions():
    score = 0
    total_asked = 0
    print("Welcome to the Capitals Quiz! Type 'quit' at any time to exit.\n")

    for question, answer in questions.items():
        user_answer = input(f"{question} ").strip()
        if user_answer.lower() == 'quit':
            print("\nQuitting the game...")
            break

        total_asked += 1
        if user_answer.lower() == answer.lower():
            print("Correct! Congratulations.\n")
            score += 1 #assignment operators
        else:
            print(f"Nope! The answer was {answer}.\n")
            score -= 1

    print(f"\nYour final score: {score}/{total_asked}")

# Run the quiz game
ask_questions()
```
